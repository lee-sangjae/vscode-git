## Count_by_range함수는 유용하게 쓸 수 있으니 꼭 기억해둘 것!

```python
def binary_search(arr, target, start, end):
    if start > end:
        return None
    
    mid = (start+end)//2

    if arr[mid] == target:
        return mid

    elif arr[mid] < target:
        return binary_search(arr, target, mid+1, end)
    else:
        return binary_search(arr, target, start, mid-1)

n, target = list(map(int, input().split()))

arr = list(map(int, input().split()))

result = binary_search(arr, target, 0, n-1)

if result == None:
    print('원소가 존재 하지 않음')
else:
    print(result+1)
```

```python
# 값이 특정 범위에 속하는 데이터 개수 구하기
from bisect import bisect_left, bisect_right

def count_by_range(a, left_v, right_v):
    right_idx = bisect_right(a, right_v)
    left_idx = bisect_left(a, left_v)
    return right_idx - left_idx


a = [1, 2, 3, 3, 3, 3, 4, 4, 8, 9]

#값이 4인 데이터 개수 출력
print(count_by_range(a, 4, 4))

#값이 [-1, 3] 범위에 있는 데이터 개수 출력 
print(count_by_range(a, -1, 3))

```

```python
n, m = map(int, input().split())

dd = list(map(int, input().split()))
```

```python
start = 0
end = max(dd)

result = 0

while(start<=end):
    
    mid = (start+end)//2

    l = sum([i-mid for i in dd]) #자른 떡 길이 총합

    if l < m: #목표치보다 떡길이 짧으면 왼쪽으로
        end = mid-1
    else: #목표치보다 떡길이 길면 오른쪽으로
        result = mid
        start = mid+1

print(result)
```

```python
# 값이 특정 범위에 속하는 데이터 개수 구하기
from bisect import bisect_left, bisect_right

def count_by_range(a, left_v, right_v):
    right_idx = bisect_right(a, right_v)
    left_idx = bisect_left(a, left_v)
    return right_idx - left_idx

n, x = map(int, input().split())

arr = list(map(int, input().split()))

#값이 [x, x] 범위에 있는 데이터 수 계산 
cnt = count_by_range(arr, x, x)

#값이 x인 원소가 존재하지 않는다면 
if cnt == 0 :
    print(-1)
else:
    print(cnt)
```
