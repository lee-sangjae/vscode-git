### Stack 구현 예제

```python
stack = []

stack.append(5)
stack.append(2)
stack.append(3)
stack.append(7)
stack.pop()
stack.append(1)
stack.append(4)
stack.pop()

print(stack[::-1]) #최상단 원소부터 출력
print(stack) #최하단 원소부터 출력 
```

### Queue 구현 예제

```python
from collections import deque  #큐 구현할 때는 리스트쓰는 것보다 deque쓰는 게 훨씬 효율적 

queue = deque()

# 삽입(5) - 삽입(2) - 삽입(3) - 삽입(7) - 삭제() - 삽입(1) - 삽입(4) - 삭제()
queue.append(5) #deque은 오른쪽에서 왼쪽으로 들어온다고 이해하면 됨
queue.append(2)
queue.append(3)
queue.append(7)
print(queue.popleft())
queue.append(1)
queue.append(4)
print(queue.popleft())

print(queue)
queue.reverse()
print(queue)
```

### 팩토리얼 구현 예제

```python
# 반복적으로 구현한 n!
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result*=i
    return result

#재귀적으로 구현한 n!
def factorial2(n):
    if n<=1:
        return 1
    return n*factorial2(n-1)

print(factorial(7))
print(factorial2(7))
```

### 유클리드 호제법

```python
def gcd(a,b):
    if a%b==0:
        return b
    else:
        return gcd(b, a%b)

print(gcd(192,162))
```

### DFS 소스코드 예제

```python
def dfs(g, v, visited):
    #현재 노드 방문 처리
    visited[v] = True
    print(v, end=' ')

    #현재 노드와 연결된 다른 노드를 재귀적으로 방문
    for i in g[v]:
        if not visited[i]:
            dfs(g, i, visited)

g= [
    [],
    [2, 3, 8],
    [1, 7],
    [1, 4, 5],
    [3, 5],
    [3, 4],
    [7],
    [2, 6, 8],
    [1, 7]
]

visited = [False] * 9

dfs(g, 1, visited)
```

### BFS 구현 예제

```python
from collections import deque

def bfs(g, start, visited):
    #큐 구현을 위해 deque 라이브러리 사용
    q = deque([start])

    #현재 노드 방문 처리
    visited[start] = True

    #큐 빌 때까지 반복

    while q:
        #큐에서 원소 하나 뽑아서 출력
        v = q.popleft()
        print(v, end = ' ')

        #아직 방문하지 않은 인접 원소들을 큐에 삽입 
        for i in g[v]:
            if not visited[i]:
                q.append(i)
                visited[i] = True

g= [
    [],
    [2, 3, 8],
    [1, 7],
    [1, 4, 5],
    [3, 5],
    [3, 4],
    [7],
    [2, 6, 8],
    [1, 7]
]

visited = [False] * 9

bfs(g, 1, visited)
```

```python
# 내 구현 

n, m = map(int, input().split())

#2차원 리스트 맵 정보 입력 받기
graph = []
for i in range(n):
    graph.append(list(map(int, input())))

g = graph.copy()

def dfs(y, x):
    #주어진 범위 벗어나면 즉시 종료
    if y<0 or y>=n or x<0 or x>=m:
        return False

    #현재 노드를 아직 방문하지 않았다면,
    if g[y][x] == 0:
        #해당노드 방문 처리
        g[y][x] = 1

        #상하좌우 위치 모두 재귀 호출
        dfs(y-1,x)
        dfs(y+1,x)
        dfs(y,x-1)
        dfs(y,x+1)
        return True #특정노드에서 시작해서 주변 구역 다 확보했으면 True가 리턴되는 구조
    return False

#모든 노드에 대하여 음료수 채우기
result = 0

for i in range(n):
    for j in range(m):
        #현재 위치에서 dfs 수행
        if dfs(i, j) == True:
            result+=1
print(result)
```

```python
import copy
n, m = map(int, input().split())

graph = []
for i in range(n):
    graph.append(list(map(int, input())))


g = copy.deepcopy(graph)

def bfs(y, x):
    dq = deque()
    dq.append((y,x))

    #큐 빌떄까지 반복
    while dq:
        y, x = dq.popleft()

        #4방향 확인
        for i in range(4):
            ny = y+dy[i]
            nx = x+dx[i]

            #벗어나면 무시
            if ny<0 or ny>=n or nx <0 or nx>=m:
                continue
            
            #괴물있어도 무시
            if g[ny][nx] == 0:
                continue

            if g[ny][nx] == 1:
                g[ny][nx] = g[y][x]+1
                dq.append((ny, nx))
    
    return g[n-1][m-1]

#상하좌우
dy = [-1, 1, 0, 0]
dx = [0, 0, -1, 1]

print(bfs(0,0))

    
```
